### Addestramento di un Semplice Modello di Regressione Lineare

Questo esempio mostra come addestrare un modello di regressione lineare utilizzando `scikit-learn`.

````python
# @title Testo del titolo predefinito
import numpy as np
import matplotlib.pyplot as plt
from sklearn.linear_model import LinearRegression
from sklearn.model_selection import train_test_split
from sklearn.metrics import mean_squared_error, r2_score

# 1. Genera Dati Sintetici
# Vogliamo prevedere 'y' da 'x', con un po' di rumore
x = np.random.rand(100, 1) * 10  # 100 punti dati tra 0 e 10
y = 2 * x + 1 + np.random.randn(100, 1) * 2 # y = 2x + 1 + rumore

# 2. Visualizza i Dati Generati
plt.figure(figsize=(8, 6))
plt.scatter(x, y, color='blue', label='Dati Generati')
plt.title('Dati Sintetici per Regressione Lineare')
plt.xlabel('Variabile Indipendente (x)')
plt.ylabel('Variabile Dipendente (y)')
plt.legend()
plt.grid(True)
plt.show()
````

````python
# 3. Dividi i Dati in Set di Addestramento e Test
# Utilizziamo l'80% dei dati per l'addestramento e il 20% per il test
X_train, X_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)

print(f"Dimensioni del set di addestramento: {X_train.shape[0]} esempi")
print(f"Dimensioni del set di test: {X_test.shape[0]} esempi")

# 4. Definisci il Modello di Regressione Lineare
model = LinearRegression()

# 5. Addestra il Modello
# Il modello impara la relazione tra X_train e y_train
model.fit(X_train, y_train)

print("\nAddestramento del modello completato.")
print(f"Coefficiente (pendenza): {model.coef_[0][0]:.2f}")
print(f"Intercetta: {model.intercept_[0]:.2f}")
````

````python
# 6. Fai Previsioni sul Set di Test
y_pred = model.predict(X_test)

# 7. Valuta il Modello
mse = mean_squared_error(y_test, y_pred)
r2 = r2_score(y_test, y_pred)

print(f"\nErrore Quadratico Medio (MSE) sul set di test: {mse:.2f}")
print(f"Coefficiente di Determinazione (R^2) sul set di test: {r2:.2f}")
````

````python
# 8. Visualizza i Risultati dell'Addestramento e le Previsioni
plt.figure(figsize=(8, 6))
plt.scatter(X_test, y_test, color='red', label='Dati di Test Reali')
plt.plot(X_test, y_pred, color='green', linewidth=3, label='Previsioni del Modello')
plt.title('Previsioni del Modello di Regressione Lineare')
plt.xlabel('Variabile Indipendente (x)')
plt.ylabel('Variabile Dipendente (y)')
plt.legend()
plt.grid(True)
plt.show()
````
